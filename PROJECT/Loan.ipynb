```python
Q1 = df["Income"].quantile(0.25)
Q3 = df["Income"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["Income"] < lower) | (df["Income"] > upper)]

outliers
```

```python
# Histogram
import matplotlib.pyplot as plt

df["Loan_amount"].hist(bins=25,edgecolor='black')

plt.title("Loan Amount Distribution")
plt.xlabel("Loan Amount")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.hist(df["age"], bins=20,color='green',edgecolor='black')
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
# Box Plot
plt.boxplot(df["Loan_amount"])

plt.title("Loan Amount Box Plot")

plt.show()
```

```python
# Bar chart
df["EmploymentType"].value_counts().plot(kind="bar")

plt.title("Employment Type")

plt.show()
```

```python
# Pie chart
df["EmploymentType"].value_counts().plot(kind="pie", autopct="%1.1f%%")

plt.ylabel("")

plt.show()
```

```python
Q1 = df['Loan_amount'].quantile(0.25)
Q3 = df['Loan_amount'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

loanamount_outliers = df[(df['Loan_amount'] < lower) | (df['Loan_amount'] > upper)]

print("LoanAmount Outliers:", len(loanamount_outliers))
```

```python
#Scatter Chart
df_sample=df.sample(100,random_state=10)
plt.scatter(df_sample['LoanTerm'],df_sample['Loan_amount'])
plt.title("Laon Term vs Loan Amount")
plt.xlabel("Loan Term")
plt.ylabel("Loan Amount")
plt.show()
```

```python
import seaborn as sns
df_sample=df.sample(1000,random_state=42)
plt.scatter(df_sample["Income"], df_sample["Loan_amount"])

plt.title("Income vs Loan Amount")
plt.xlabel("Income")
plt.ylabel("Loan Amount")
plt.show()
```

```python
#Bar chart
pd.crosstab(df["EmploymentType"], df["Default"]).plot(kind="bar")

plt.title("Default by Employment Type")
plt.xlabel("Employment Type")
plt.ylabel("Count")
plt.show()
```

```python
pd.crosstab(df["Education"], df["Default"]).plot(kind="bar",color=['red','green'])

plt.title("Default by Education")
plt.xlabel("Education")
plt.ylabel("Count")
plt.show()
```
