PREPROCESSING DATA

```python
import matplotlib.pyplot as plt
```

```python
Q1 = df["Income"].quantile(0.25)
Q3 = df["Income"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["Income"] < lower) | (df["Income"] > upper)]

outliers
```

```python
df["Default"].value_counts().plot(kind='bar')
plt.title("Loan Default Distribution")
plt.xlabel("Default")
plt.ylabel("Count")
plt.show()
```

```python
plt.hist(df["Age"], bins=20,color='green',edgecolor='black')
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.hist(df["Income"], bins=25,color='red',edgecolor='black')

plt.title("Income Distribution")
plt.xlabel("Income")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.hist(df["CreditScore"], bins=25,color='yellow',edgecolor='lightblue')

plt.title("Credit Score Distribution")
plt.xlabel("Credit Score")
plt.ylabel("Frequency")
plt.show()
```

```python
import seaborn as sns

plt.figure(figsize=(12,8))

sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap="coolwarm")

plt.title("Correlation Matrix")
plt.show()
```

```python
plt.boxplot(x=df["Income"] ,vert=False)
plt.title("Income Boxplot")
plt.show()
```

```python
sns.boxplot(x=df["LoanAmount"])

plt.title("Loan Amount Boxplot")
plt.show()
```

```python
pd.crosstab(df["EmploymentType"], df["Default"]).plot(kind="bar")

plt.title("Default by Employment Type")
plt.xlabel("Employment Type")
plt.ylabel("Count")
plt.show()
```

```python
pd.crosstab(df["Education"], df["Default"]).plot(kind="bar",color=['red','green'])

plt.title("Default by Education")
plt.xlabel("Education")
plt.ylabel("Count")
plt.show()
```

```python
df_sample=df.sample(1000,random_state=42)
plt.scatter(df_sample["Income"], df_sample["LoanAmount"])

plt.title("Income vs Loan Amount")
plt.xlabel("Income")
plt.ylabel("Loan Amount")
plt.show()
```

```python
df_sample=df.sample(100,random_state=10)
plt.scatter(df_sample['LoanTerm'],df_sample['LoanAmount'])
plt.title("Laon Term vs Loan Amount")
plt.xlabel("Loan Term")
plt.ylabel("Loan Amount")
plt.show()
```
